```python
import pandas as pd
from sklearn.datasets import load_iris

# 1. Load the dataset
iris = load_iris()

# 2. Define the DataFrame 'df' FIRST
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# 3. Add the species names column
df['species'] = iris.target_names[iris.target]

# 4. Display the first 5 rows
df.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

# Loop through unique species
for species in df["species"].unique():
    subset = df[df["species"] == species]
    plt.scatter(
        subset["petal length (cm)"],
        subset["petal width (cm)"],
        label=species
    )

# Formatting (moved outside the loop so it only runs once)
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.title("Iris species based on petal measurements")
plt.legend()

plt.show()
```
